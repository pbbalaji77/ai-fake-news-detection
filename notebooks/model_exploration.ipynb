# AI-Powered Fake News Detection System: Exploratory Data Analysis (EDA)

**Objective**:
Perform comprehensive Exploratory Data Analysis on news articles to understand class distributions, length distributions, vocabulary richness, and frequent n-grams in REAL versus FAKE articles.

```python
import sys
from pathlib import Path

# Set up project path
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from src.data_loader import load_raw_data, clean_dataset, split_dataset
from src.eda import compute_class_distribution, compute_text_length_statistics, extract_top_ngrams
from src.config import COL_LABEL, COL_CLEAN_TEXT, LABEL_MAP, LABEL_REAL, LABEL_FAKE

print(f"Project root resolved: {PROJECT_ROOT}")
```

## 1. Load and Inspect Dataset

```python
raw_df = load_raw_data()
print(f"Raw dataset shape: {raw_df.shape}")
raw_df.head()
```

## 2. Clean and Normalize Dataset

```python
cleaned_df = clean_dataset(raw_df)
print(f"Cleaned dataset shape: {cleaned_df.shape}")
cleaned_df[['title', 'label', 'clean_text']].head()
```

## 3. Class Balance Analysis

```python
class_stats = compute_class_distribution(cleaned_df)
print("Class Distribution:", class_stats)

plt.figure(figsize=(6, 4))
sns.countplot(data=cleaned_df, x='label', palette=['#2ecc71', '#e74c3c'])
plt.title("Class Distribution (0: REAL, 1: FAKE)")
plt.show()
```

## 4. Text Length and Word Count Analysis

```python
length_stats = compute_text_length_statistics(cleaned_df)
print("Length Statistics:", length_stats)

cleaned_df['word_count'] = cleaned_df['clean_text'].apply(lambda x: len(str(x).split()))
plt.figure(figsize=(8, 4))
sns.boxplot(data=cleaned_df, x='label', y='word_count', palette=['#2ecc71', '#e74c3c'])
plt.title("Word Count Distribution by Class")
plt.show()
```

## 5. N-Gram & Vocabulary Analysis

```python
real_texts = cleaned_df[cleaned_df[COL_LABEL] == LABEL_REAL][COL_CLEAN_TEXT].tolist()
fake_texts = cleaned_df[cleaned_df[COL_LABEL] == LABEL_FAKE][COL_CLEAN_TEXT].tolist()

print("Top Unigrams in REAL:", extract_top_ngrams(real_texts, n=1, top_k=5))
print("Top Unigrams in FAKE:", extract_top_ngrams(fake_texts, n=1, top_k=5))
print("Top Bigrams in REAL:", extract_top_ngrams(real_texts, n=2, top_k=3))
print("Top Bigrams in FAKE:", extract_top_ngrams(fake_texts, n=2, top_k=3))
```
